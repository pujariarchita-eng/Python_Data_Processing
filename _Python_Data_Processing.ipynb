# Python Data Processing

## Project Description

This project demonstrates basic data processing using Python.
Student records are stored using lists and dictionaries, and useful
summary statistics are calculated from the records.

## Objective

To process structured records using Python and generate useful
summary statistics.

```python
# Student records

students = [
    {"name": "Aarav", "marks": 85, "age": 20, "department": "ENTC"},
    {"name": "Priya", "marks": 92, "age": 21, "department": "Computer"},
    {"name": "Rahul", "marks": 76, "age": 20, "department": "Mechanical"},
    {"name": "Sneha", "marks": 88, "age": 22, "department": "ENTC"},
    {"name": "Aditya", "marks": 67, "age": 21, "department": "Civil"},
    {"name": "Ananya", "marks": 95, "age": 20, "department": "Computer"},
    {"name": "Rohan", "marks": 72, "age": 22, "department": "Mechanical"},
    {"name": "Kavya", "marks": 81, "age": 21, "department": "ENTC"}
]

students
```

```python
# Display student records

for student in students:
    print(
        f"Name: {student['name']}, "
        f"Marks: {student['marks']}, "
        f"Age: {student['age']}, "
        f"Department: {student['department']}"
    )
```

```python
# Function to calculate summary statistics

def calculate_statistics(students):
    marks = [student["marks"] for student in students]
    ages = [student["age"] for student in students]

    statistics = {
        "Total Students": len(students),
        "Average Marks": sum(marks) / len(marks),
        "Highest Marks": max(marks),
        "Lowest Marks": min(marks),
        "Average Age": sum(ages) / len(ages)
    }

    return statistics
```

```python
# Generate summary statistics

summary = calculate_statistics(students)

for key, value in summary.items():
    if isinstance(value, float):
        print(f"{key}: {value:.2f}")
    else:
        print(f"{key}: {value}")
```

```python
# Filter students with marks 80 or above

high_scorers = [
    student for student in students
    if student["marks"] >= 80
]

print("Students scoring 80 or above:")

for student in high_scorers:
    print(f"{student['name']} - {student['marks']} marks")
```

```python
# Classify students as Pass or Fail

for student in students:
    if student["marks"] >= 40:
        student["result"] = "Pass"
    else:
        student["result"] = "Fail"

print("Student Results:")

for student in students:
    print(f"{student['name']} - {student['marks']} marks - {student['result']}")
```

```python
# Count pass and fail students

pass_count = sum(1 for student in students if student["result"] == "Pass")
fail_count = sum(1 for student in students if student["result"] == "Fail")

print("Number of Pass Students:", pass_count)
print("Number of Fail Students:", fail_count)
```

```python
# Count students by department

department_count = {}

for student in students:
    department = student["department"]
    department_count[department] = department_count.get(department, 0) + 1

print("Students by Department:")

for department, count in department_count.items():
    print(f"{department}: {count}")
```
